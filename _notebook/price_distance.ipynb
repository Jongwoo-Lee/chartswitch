```python
import pandas as pd
import random

df = pd.DataFrame(columns=['event_ts', 'close_price'])

event_ts = pd.to_datetime('2024-05-01 00:00:00.000', format='%Y-%m-%d %H:%M:%S.%f')
close_price = float(10000.0)

new_df = pd.DataFrame([{'event_ts': event_ts, 'close_price': close_price}])
if df.empty:
  df = new_df
else:
  df = pd.concat([df, new_df], ignore_index=True)

i = 0
randsum = 0
while i < 1000:
  i += 1
  rand = random.randint(-9950, 10000) / 500000
  randsum += rand
  event_ts += pd.Timedelta(seconds=1)
  close_price *= 1 + rand
  new_df = pd.DataFrame([{'event_ts': event_ts, 'close_price': close_price}])
  df = pd.concat([df, new_df], ignore_index=True)
```

```python

# Sort the DataFrame by event_ts
df = df.sort_values(by='event_ts')

# Calculate percentage change
df['pct_change'] = df['close_price'].pct_change()

# Calculate absolute percentage change
df['abs_pct_change'] = df['pct_change'].abs()

# Calculate the rolling sum of the last 100 absolute percentage changes
# Note: rolling window size of 100, use min_periods=1 to start calculating immediately
df['rolling_fluctuation'] = df['abs_pct_change'].rolling(window=100, min_periods=1).sum()

# Multiply by 100 to express as a percentage
df['rolling_fluctuation'] *= 100
df
```
