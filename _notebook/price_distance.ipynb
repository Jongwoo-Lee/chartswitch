```python
import pandas as pd
import random

price_data = pd.DataFrame(columns=['event_ts', 'close_price'])

event_ts = pd.to_datetime('2024-05-01 00:00:00.000', format='%Y-%m-%d %H:%M:%S.%f')
close_price = float(10000.0)

new_df = pd.DataFrame([{'event_ts': event_ts, 'close_price': close_price}])
if price_data.empty:
  price_data = new_df
else:
  price_data = pd.concat([price_data, new_df], ignore_index=True)

i = 0
while i < 1000:
  i += 1
  rand = random.randint(-9950, 10000) / 500000
  event_ts += pd.Timedelta(seconds=1)
  close_price *= 1 + rand
  new_df = pd.DataFrame([{'event_ts': event_ts, 'close_price': close_price}])
  price_data = pd.concat([price_data, new_df], ignore_index=True)
  
price_data
```
